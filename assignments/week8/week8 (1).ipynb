```python
#BAI 9
from datetime import datetime
ngay_sinh = input("Nhap ngay sinh (y/m/d): ")
ngay_sinh = datetime.strptime(ngay_sinh, "%Y/%m/%d")
hom_nay = datetime.today()
tuoi = hom_nay.year - ngay_sinh.year

if (hom_nay.month, hom_nay.day) < (ngay_sinh.month, ngay_sinh.day):
    tuoi -= 1

print("Tuoi cua ban la:", tuoi)
```

```python
#BAI 10
gio = float(input("Nhap so gio lam: "))
luong = float(input("Nhap tien luong moi gio: "))

if gio <= 44:
    tien = gio * luong
else:
    tien = 44 * luong + (gio - 44) * luong * 1.5

print("Tien luong:", tien)
```

```python
#BAI 11
import math
C = 50
H = 30

D = input("Nhap cac gia tri D: ").split(",")
ket_qua = []

for d in D:
    Q = math.sqrt((2 * C * int(d)) / H)
    ket_qua.append(str(round(Q)))

print(",".join(ket_qua))
```

```python
#BAI 22
so_du=0
print("Nhập D số_tiền hoặc W số_tiền; Enter trống để kết thúc")
while True:
    s=input("Giao dịch: ").strip()
    if s=="":
        break
    a=s.split()
    if len(a)!=2:
        print("Giao dịch không hợp lệ")
        continue
    loai=a[0].upper()
    try:
        tien=int(a[1])
    except ValueError:
        print("Số tiền không hợp lệ")
        continue
    if tien<0:
        print("Số tiền phải không âm")
    elif loai=="D":
        so_du+=tien
    elif loai=="W":
        so_du-=tien
    else:
        print("Chỉ nhận D hoặc W")
print("Số dư:",so_du)
```

```python
#BAI 23
def kiem_tra_doi_xung(a):
    return a==a[::-1]

def sx_doi_xung(a):
    b=a.copy()
    for i in range(len(b)//2):
        b[-i-1]=b[i]
    return b

s=input("Nhập dãy số nguyên cách nhau bởi dấu phẩy: ")
a=[int(x.strip()) for x in s.split(",")]
if kiem_tra_doi_xung(a):
    print("Dãy đã đối xứng:",a)
else:
    print("Dãy ban đầu chưa đối xứng")
    b=sx_doi_xung(a)
    print("Dãy sau khi biến đổi:",b)
    print("Kiểm tra đối xứng:",kiem_tra_doi_xung(b))
```
