```python
# Bai 14 - Tam giác 
def tamgiac():
    a, b, c = map(float, input("Nhap 3 canh cua tam giac: ").split())
    if a + b > c and a + c > b and b + c > a:
        print("la 1 tam giac")
        p = (a + b + c) / 2
        S = (p * (p - a) * (p - b) * (p - c)) ** 0.5
        print("Dien tich tam giac la:", S)
        return S
    else:
        print("khong la tam giac")
        return a, b, c
tamgiac()
```

```python
# Bai 20 - Phuong trinh bac 2
def phuongtrinh():
    a = float(input("Nhap he so a: "))
    b = float(input("Nhap he so b: "))
    c = float(input("Nhap he so c: "))
    delta = b ** 2 - 4 * a * c
    if delta < 0:
        print("Phuong trinh vo nghiem")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"Phuong trinh co nghiem kep: {x}")
    else:
        x1 = (-b + delta ** 0.5) / (2 * a)
        x2 = (-b - delta ** 0.5) / (2 * a)
        print(f"Phuong trinh co hai nghiem phan biet: x1 = {x1}, x2 = {x2}")
    return delta
phuongtrinh()
```

```python
#Bài 32:Giải hệ phương trình bậc nhất nhất 2 ẩn
print('Hệ phương trình có dạng:')
print('''    ax + by = m 
             cx + dy = n''')
def he():
  a = int(input('Nhập hệ số a: '))
  b = int(input('Nhập hệ số b: '))
  m = int(input('Nhập hệ số m: '))
  c = int(input('Nhập hệ số c: '))
  d = int(input('Nhập hệ số d: '))
  n = int(input('Nhập hệ số n: '))
  D = a*d-b*c
  Dx = m*d-b*n
  Dy = a*n-c*m
  if D != 0:
    print('Hệ phương trình có nghiệm là: x = ', Dx/D, 'y = ', Dy/D)
  elif D == 0:
    if Dx == Dy ==0:
      print('Hệ phương trình vô số nghiệm')
    else:
      print('Hệ phương trình vô nghiệm')
he()
```

```python
#Bài 33: Điểm thuộc đường tròn:
def duong_tron():
  r = float(input('Nhập bán kính r = '))
  print('Nhập tọa độ tâm đường tròn:')
  x_0 = float(input('x0 = '))
  y_0 = float(input('y0 = '))
  print('Nhập tọa độ điểm A')
  x_a = float(input('xa = '))
  y_a = float(input('yb = '))
  a = x_0 - x_a
  b = y_0 - y_a
  d = (a**2+b**2)**0.5
  if d < r:
    print('Điểm nằm trong đường tròn')
  elif d == r:
    print('Điểm nằm trên đường tròn')
  else:
    print('Điểm nằm ngoài đường tròn')
duong_tron()
```

```python
#Bài 49: Số Armstrong
def armstrong():
    for n in range(100, 1000):
        tram = n // 100
        chuc = (n // 10) % 10
        don_vi = n % 10

        if n == tram**3 + chuc**3 + don_vi**3:
            print(n)
armstrong()
```
